Now generate the heatmap for number of cells in each condition 
-

```python
def plot_drug_combo_cell_counts(
    df,
    trametinib_col="trametinib_dose",
    treatment_col="treatment",
    dose_col="dose",
    treatment_order=None,
    trametinib_order=None,
    cmap="YlGnBu",
    figsize=None,
    annotate=True,
    ax=None,
):
    """
    Plot the number of cells observed for each trametinib × partner-drug dose combination.

    Parameters
    ----------
    df : pd.DataFrame
        One row per cell.
    trametinib_col, treatment_col, dose_col : str
        Column names containing trametinib dose, partner drug, and partner-drug dose.
    treatment_order : list, optional
        Desired left-to-right order of partner drugs. Defaults to alphabetical order.
    trametinib_order : list, optional
        Desired top-to-bottom order of trametinib doses. Defaults to ascending numeric order.
    cmap : str
        Seaborn/matplotlib colormap.
    figsize : tuple, optional
        Figure size. Automatically chosen if omitted.
    annotate : bool
        Add cell-count labels to each heatmap tile.
    ax : matplotlib.axes.Axes, optional
        Existing axis to plot into.

    Returns
    -------
    fig, ax, count_matrix
    """

    plot_df = df[[trametinib_col, treatment_col, dose_col]].copy()

    # Ensure dose columns are numeric.
    plot_df[trametinib_col] = pd.to_numeric(plot_df[trametinib_col])
    plot_df[dose_col] = pd.to_numeric(plot_df[dose_col])

    if trametinib_order is None:
        trametinib_order = sorted(plot_df[trametinib_col].dropna().unique())

    if treatment_order is None:
        treatment_order = sorted(plot_df[treatment_col].dropna().unique())

    # Get all doses observed for each partner drug and construct ordered heatmap columns.
    column_pairs = []
    treatment_boundaries = []
    running_total = 0

    for treatment in treatment_order:
        doses = sorted(
            plot_df.loc[
                plot_df[treatment_col].eq(treatment), dose_col
            ].dropna().unique()
        )
        column_pairs.extend([(treatment, dose) for dose in doses])

        running_total += len(doses)
        treatment_boundaries.append(running_total)

    # Count cells and include zero-count dose combinations.
    counts = (
        plot_df.groupby([trametinib_col, treatment_col, dose_col])
        .size()
        .rename("n_cells")
        .reset_index()
    )

    full_index = pd.MultiIndex.from_product(
        [trametinib_order, column_pairs],
        names=[trametinib_col, "treatment_dose"]
    )

    count_matrix = (
        counts.set_index([trametinib_col, treatment_col, dose_col])["n_cells"]
        .reindex(
            pd.MultiIndex.from_tuples(
                [(tram, treatment, dose)
                 for tram in trametinib_order
                 for treatment, dose in column_pairs],
                names=[trametinib_col, treatment_col, dose_col]
            ),
            fill_value=0,
        )
        .to_numpy()
        .reshape(len(trametinib_order), len(column_pairs))
    )

    count_matrix = pd.DataFrame(
        count_matrix,
        index=trametinib_order,
        columns=pd.MultiIndex.from_tuples(
            column_pairs, names=["Treatment", "Dose (µM)"]
        ),
    )

    if figsize is None:
        figsize = (
            max(10, 0.72 * len(column_pairs) + 2),
            max(4, 0.65 * len(trametinib_order) + 2),
        )

    if ax is None:
        fig, ax = plt.subplots(figsize=figsize)
    else:
        fig = ax.figure

    sns.heatmap(
        count_matrix,
        cmap=cmap,
        annot=annotate,
        fmt=",d",
        linewidths=0.5,
        linecolor="white",
        cbar_kws={
            "label": "Number of cells",
            "pad": 0.015,      # distance between heatmap and colorbar
            "shrink": 0.92,
        },
        annot_kws={"fontsize": 9},
        ax=ax,
    )

    # Bottom labels: partner-drug dose for each column.
    ax.set_xticklabels(
        [f"{dose:g}" for _, dose in column_pairs],
        rotation=0,
        fontsize=9,
    )
    ax.set_yticklabels(
        [f"{dose:g}" for dose in trametinib_order],
        rotation=0,
        fontsize=10,
    )

    ax.set_xlabel("Partner-drug dose (µM)", fontsize=11, labelpad=10)
    ax.set_ylabel("Trametinib dose (µM)", fontsize=11, labelpad=10)

    # Black separators between treatment groups.
    for boundary in treatment_boundaries[:-1]:
        ax.axvline(boundary, color="black", linewidth=2)

    # Group labels above each set of partner-drug doses.
    start = 0
    for treatment in treatment_order:
        n_doses = sum(t == treatment for t, _ in column_pairs)
        midpoint = start + n_doses / 2

        ax.text(
            midpoint,
            -0.16,
            treatment,
            ha="center",
            va="bottom",
            fontsize=11,
            transform=ax.transData,
        )
        start += n_doses

    ax.set_title(
        "Cell counts across trametinib–partner drug dose combinations",
        fontsize=13,
        fontweight="bold",
        pad=30,
    )

    sns.despine(left=True, bottom=True)
    fig.tight_layout()

    return fig, ax, count_matrix
```

```python
data = pd.read_csv('STATE/A172_qc_log1p_obs.csv', index_col=0)
columns_keep = ['cell_type', 'num_genes_expressed', 'trametinib_dose', 'treatment', 'dose', 'dose_character']
data = data[columns_keep]
data = data[data['treatment'].isin(seen_drugs)]
data
```

```python
fig, ax, count_matrix = plot_drug_combo_cell_counts(
    data,
    figsize=(10, 3),
    treatment_order=[
        "Doxorubicin",
        "Temsirolimus",
        "Palbociclib",
        "Infigratinib",
        "Volasertib",
    ],
)
fig.savefig('STATE/figures/A172_cell_counts_heatmap_seen_drugs.png', dpi=300)
```
